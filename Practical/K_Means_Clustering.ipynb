1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

2. Define and Initialize Dataset

```python
# Sample dataset
X = np.array([
    [1, 2],
    [1, 4],
    [2, 3],
    [2, 5],
    [8, 7],
    [9, 8],
    [8, 9],
    [10, 8],
    [4, 1],
    [5, 2],
    [4, 3],
    [5, 1]
])

print("Dataset:")
print(X)
```

3. Apply K-Means Clustering

```python
# Define number of clusters
k = 3

# Initialize K-Means model
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

# Fit the model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

# Get cluster centers
centers = kmeans.cluster_centers_

print("Cluster Labels:")
print(labels)

print("\nCluster Centers:")
print(centers)
```

4. Display Data with Cluster Labels

```python
for i in range(len(X)):
    print(f"Point {X[i]} -> Cluster {labels[i]}")
```

```python
5. Visualize K-Means Clustering
```

```python
plt.figure(figsize=(8, 6))

# Plot data points
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    s=100
)

# Plot cluster centers
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker='X',
    s=250,
    edgecolors='black'
)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)

plt.show()
```
